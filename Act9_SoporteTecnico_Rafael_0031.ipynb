```python
# ============================================
# SIMULADOR DE INTELIGENCIA ARTIFICIAL
# Tema: Soporte Técnico
# Uso de if, elif, else y operadores
# Sin funciones, sin ciclos for y sin POO
# ============================================

soportetecnico = []

print("====================================")
print(" IA DE SOPORTE TECNICO ")
print("====================================")

while True:

    print("\nHola tecnico.")
    nombre = input("Ingresa tu nombre: ")

    print("\nSelecciona el tipo de problema")
    print("1. Equipo lento")
    print("2. Virus detectado")
    print("3. Equipo no enciende")

    problema = input("Opcion: ")

    # ----------------------------------
    # EQUIPO LENTO
    # ----------------------------------
    if problema == "1":

        print("\nProblema seleccionado: Equipo Lento")

        memoria = int(input("Memoria RAM disponible (GB): "))
        programas = int(input("Cantidad de programas abiertos: "))

        if memoria >= 8 and programas <= 5:
            print("\nIA:", nombre + ", el equipo funciona correctamente.")
            print("No se detectan problemas graves.")
            soportetecnico.append("Equipo Lento - Estable")

        elif memoria >= 4 and programas <= 10:
            print("\nIA: El equipo presenta lentitud moderada.")
            print("Se recomienda cerrar aplicaciones.")
            soportetecnico.append("Equipo Lento - Moderado")

        else:
            print("\nIA: El equipo esta muy lento.")
            print("Se recomienda ampliar memoria RAM.")
            soportetecnico.append("Equipo Lento - Critico")

    # ----------------------------------
    # VIRUS DETECTADO
    # ----------------------------------
    elif problema == "2":

        print("\nProblema seleccionado: Virus")

        antivirus = input("¿Tiene antivirus instalado? (si/no): ")
        amenazas = int(input("Cantidad de amenazas detectadas: "))

        if antivirus == "si" and amenazas == 0:
            print("\nIA: El sistema esta protegido.")
            soportetecnico.append("Virus - Protegido")

        elif antivirus == "si" and amenazas <= 5:
            print("\nIA: Existen amenazas menores.")
            print("Realiza un escaneo completo.")
            soportetecnico.append("Virus - Riesgo Medio")

        else:
            print("\nIA: Riesgo alto de infeccion.")
            print("Instala o actualiza el antivirus.")
            soportetecnico.append("Virus - Riesgo Alto")

    # ----------------------------------
    # NO ENCIENDE
    # ----------------------------------
    elif problema == "3":

        print("\nProblema seleccionado: No Enciende")

        bateria = int(input("Nivel de bateria (%): "))
        cargador = input("¿El cargador funciona? (si/no): ")

        if bateria >= 50 and cargador == "si":
            print("\nIA: El equipo deberia encender correctamente.")
            soportetecnico.append("No Enciende - Revisar Sistema")

        elif bateria >= 20 or cargador == "si":
            print("\nIA: Posible problema de energia.")
            soportetecnico.append("No Enciende - Energia")

        else:
            print("\nIA: Falla critica detectada.")
            print("Se recomienda revision tecnica.")
            soportetecnico.append("No Enciende - Critico")

    # ----------------------------------
    # OPCION INVALIDA
    # ----------------------------------
    else:

        print("\nIA: Opcion no valida.")
        print("Selecciona una opcion correcta.")
        soportetecnico.append("Opcion Invalida")

    print("\nHistorial de diagnosticos:")
    print(soportetecnico)

    continuar = input("\n¿Deseas realizar otro diagnostico? (si/no): ")

    if continuar.lower() != "si":
        break

print("\nFin de la simulacion.")
```
