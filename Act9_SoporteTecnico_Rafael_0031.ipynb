```python
# Act. 9
# SoporteTecnico
# Arevalo Rafael
# Fecha: 25/May/26
# NL = 3

soportetecnico = []   # Lista vacía

print("================================")
print("   IA DE SOPORTE TECNICO")
print("================================")

while True:

    print("\n------ MENU ------")
    print("1. Registrar soporte")
    print("2. Revisar soportes")
    print("3. Diagnosticar")
    print("4. Salir")
    print("------------------")

    op = input("Selecciona una opcion: ")

    if op == "1":

        equipo = input("Nombre del equipo: ")
        soportetecnico.append(equipo)
        print("Equipo registrado")

    elif op == "2":

        if soportetecnico == []:
            print("Lista vacia")
        else:
            print(soportetecnico)

    elif op == "3":

        falla = input("Problema (lento/virus/no enciende): ")

        if falla == "lento":
            print("Limpiar archivos")

        elif falla == "virus":
            print("Instalar antivirus")

        elif falla == "no enciende":
            print("Revisar cargador")

        else:
            print("Problema no identificado")

    elif op == "4":
        print("Programa terminado")
        break

    else:
        print("Opcion incorrecta")
```
