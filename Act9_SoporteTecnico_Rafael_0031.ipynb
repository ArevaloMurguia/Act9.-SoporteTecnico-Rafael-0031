```python
# Arevalo Rafael
# Fecha: 25/May/26
# NL = 3
# SoporteTecnico

soportestecnicos = []       # Lista vacía
lossoportestecnicos = {}    # Diccionario vacío

print("====================================")
print("     IA DE SOPORTE TECNICO")
print("====================================")

while True:

    print("\n----------- MENU -----------")
    print("1. Registrar soporte")
    print("2. Revisar soportes")
    print("3. Diagnosticar problema")
    print("4. Mostrar ultimo registro")
    print("5. Salir")
    print("----------------------------")

    opcion = input("Selecciona una opcion: ")

    # Registrar soporte
    if opcion == "1":

        print("\n--- REGISTRAR SOPORTE ---")
        nombre = input("Nombre del equipo: ")
        problema = input("Problema detectado: ")
        bateria = int(input("Nivel de bateria: "))
        internet = input("Internet (si/no): ")

        lossoportestecnicos = {
            "nombre": nombre,
            "problema": problema,
            "bateria": bateria,
            "internet": internet
        }

        soportestecnicos.append(lossoportestecnicos)

        print("Soporte registrado correctamente")

    # Revisar soportes
    elif opcion == "2":

        print("\n--- SOPORTES REGISTRADOS ---")

        if soportestecnicos == []:
            print("No hay soportes registrados")
        else:
            print(soportestecnicos)

    # Diagnóstico IA
    elif opcion == "3":

        print("\n--- DIAGNOSTICO IA ---")
        falla = input("Problema (lento/virus/pantalla azul/no enciende): ")
        bateria = int(input("Nivel de bateria: "))
        internet = input("Internet (si/no): ")

        if bateria < 20:
            print("IA: Bateria baja")

        if internet == "no":
            print("IA: Revisar modem")

        if falla == "lento":
            print("IA: Limpiar archivos temporales")

        elif falla == "virus":
            print("IA: Instalar antivirus")

        elif falla == "pantalla azul":
            print("IA: Reiniciar sistema")

        elif falla == "no enciende":
            print("IA: Revisar cargador")

        else:
            print("IA: Problema no identificado")

    # Mostrar diccionario
    elif opcion == "4":

        print("\n--- ULTIMO REGISTRO ---")

        if lossoportestecnicos == {}:
            print("Diccionario vacio")
        else:
            print(lossoportestecnicos)

    # Salir
    elif opcion == "5":

        print("\nPrograma terminado")
        break

    # Error
    else:
        print("Opcion incorrecta")
```
